Alg 01: Swapping values from one variable to the other
<br>
Basically we are trying to implement the concept of **Time Space Complexity Trade Off**.

```python
var1 =5
var2 =10

#we are creating a new variable named as temp variable,
# which will be holding the value. the above variables vice versa

temp = var1
var1 = var2
var2 = temp

print("var1 has :",var1)
print("var2 has :",var2)
```

```python
# Now we are goin to do it without creating a third variable

var1 = 20
var2 = 12

var1 = var1 + var2

var2 = var1 - var2
var1 = var1 - var2

print("var1 has:", var1)

print("var2 has:", var2)
```

**Time Space Complexity Trade Off** states that if you want make your algorithm **time effiecient** than you'll have to *increase *or you'll *need more space*, and if you want to make your algorithm space efficient than the time need to be increased. These both function as vice versa.
