# Rcursive Algorithms - Factorial

## Given Algorithm 
- INPUT = n 
- OUTPUT = n!

```
fact(n) {
    if n == 0
        return 1
    else
        return n * fact(n-1)
    end if
}
```

## Algorithm Implementation

````python
def fact(n):
    if n == 0:
        return 1
    else:
        return n * fact(n - 1)


# Examples
print("0! =", fact(0))
print("5! =", fact(5))
print("7! =", fact(7))
````

## Time Complexity Derivation

In the recursive case the algorithm performs **three constant-time steps**:

1. the comparison `n == 0`
2. the subtraction that forms the argument `n-1`
3. the multiplication `n * fact(n-1)`

Therefore the recurrence is:

$$
T(n) =
\begin{cases}
1 & \text{if } n = 0 \\
T(n-1) + 3 & \text{if } n > 0
\end{cases}
$$

### Solving the recurrence

Unrolling:

\begin{align*}
T(n)
& = T(n-1) + 3 \\
& = T(n-2) + 3 + 3 \\
& = T(n-3) + 3 + 3 + 3 \\
& \;\vdots \\
& = T(0) + 3n \\
& = 1 + 3n
\end{align*}

**Closed form:**

$$
T(n) = 3n + 1
$$

**Asymptotic complexity:**

$$
T(n) = \Theta(n)
$$
