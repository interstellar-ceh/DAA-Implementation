# GCD of two numbers

ALGORITHM

- Step1:Take two numbers as input.
- Step2:Find the greater of the two numbers.
- Step3:Keep on dividing the greater number by the smaller number until remainder becomes 0.
- Step4:When the remainder becomes 0 print the smaller number i.e., the denominator as the output.
- Step5:Exit

TIME COMPLEXITY

- Time Complexity: O(min(a, b))

```python
# Find GCD by Applying Euclid's Formula

# Function to return GCD of a and b
def gcd(a, b):
    if a == 0:
        return b
    return gcd(b % a, a)


# Driver program
a = 10
b = 15

print(f"GCD({a}, {b}) = {gcd(a, b)}")

a = 35
b = 10

print(f"GCD({a}, {b}) = {gcd(a, b)}")
```

# with user input
If we are required to take two numbers as input, we can use:

```python
# Find GCD using Euclidean Algorithm

def gcd(a, b):
    if a == 0:
        return b
    return gcd(b % a, a)


# Take two numbers as input
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

# Find and display GCD
print(f"GCD({a}, {b}) = {gcd(a, b)}")
```
