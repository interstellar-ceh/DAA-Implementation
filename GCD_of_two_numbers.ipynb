```python
# Python Program to Find GCD by Applying Euclid's Formula

# Function to return GCD of a and b
def gcd(a, b):
    if a == 0:
        return b
    return gcd(b % a, a)


# Driver program
a = 10
b = 15

print(f"GCD({a}, {b}) = {gcd(a, b)}")

a = 35
b = 10

print(f"GCD({a}, {b}) = {gcd(a, b)}")
```
